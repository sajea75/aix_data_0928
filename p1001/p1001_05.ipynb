```python
import pandas as pd
df = pd.read_excel('file/score.xlsx',index_col='지원번호')
```

```python
import random
df['요일'] = 0

for i in range(8):
    df.iloc[i,9] = random.randint(0,6)
```

```python
# 0-일요일,1-월요일,2-화요일,3-수요일,4-목요일,5-금요일,6-토요일
# 출력하시오.
# (함수 적용)
def cal_change(x):
    if x == 0: return '일요일'
    elif x == 1: return '월요일'
    elif x ==2: return '화요일'
    elif x ==3: return '수요일'
    elif x ==4: return '목요일'
    elif x ==5: return '금요일'
    else: return '토요일'

df['요일'] = df['요일'].apply(cal_change)
```

```python
data = {
    '번호':['1번','2번','3번','4번','5번','6번','7번','8번'],
    '이름':['홍길동','유관순','이순신','강감찬','김구','김유신','홍길자','홍길순'],
    '달러':['$19,700','$18,400','$16,800','$18,700','$18,800','$20,200','$18,800','$19,000'],
    '우대':[90,40,80,40,15,80,55,100]
}

df = pd.DataFrame(data)
df.set_index('번호',inplace=True)
```

```python
df['금액'] = 0 # 컬럼추가

# 금액컬럼을 추가 하시오. ( 금액 : 달러 + 달러 x 우대 x 0.01 )
# 우대 50이상인 사람에게 달러+달러X우대x0.01
# 우대 50미만 달러

# 달러 -> int변경($,쉼표) 표시 제거
# str.replace
df['달러'] = df['달러'].str.replace('$','')
df['달러'] = df['달러'].str.replace(',','').str.strip().astype(int)
```

```python
# 2개가 넘어와야 함. df['달러'],df['우대']
def cal_change(df2):
    if df2['우대']>=50:
        return df2['달러'] + (df2['달러'] * df2['우대'] * 0.01)
    else:
        return df2['달러']    

df['금액'] = df.apply(cal_change,axis=1)
```
