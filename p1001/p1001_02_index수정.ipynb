```python
import pandas as pd
df = pd.read_excel('file/score.xlsx',index_col='지원번호')
# df = pd.read_excel('file/students.xlsx',index_col='번호')
```

```python
# 합계,평균,결과 컬럼을 추가
# 결과 평균점수가 70점이상은 합격, 불합격으로 출력하시오.
df['합계'] = df['국어'] + df['영어'] + df['수학']
df['평균'] = df['합계'] / 3
df['결과'] = '불합격'
df.loc[df['평균'] >= 70, '결과'] = '합격'
```

```python
# 컬럼순서변경
# df2 = df[['이름','평균','결과']]
# df2
df = df[['이름','평균','결과','학교','학년', '키', '국어', '영어', '수학', 'SW특기', '합계']]
```

```python
# [ cols 리스트 순서를 변경 ]

# 1.리스트 타입으로 변경후 순서를 변경
cols = list(df.columns) 

# cols[-2:] # 평균, 결과 컬럼
# type(cols[0]) # str -> list타입으로 변경하려면 []로 묶어야함
# type(cols[1:-2]) # list 이름 ~ 합계
# 2. 리스트 순서 변경
cols = [cols[0]]+cols[-2:]+cols[1:-2] # str + list + list => error

# 3. 변경된 리스트 적용
df = df[cols]
```

```python
# 컬럼수정 - replace,존재하는 컬럼에 값을 입력하면 수정됨
# 컬럼추가 - 없는 컬럼에 값을 입력하면 추가됨
# 컬럼삭제 - drop : 여러개 삭제 가능, inplace=True 적용
```
